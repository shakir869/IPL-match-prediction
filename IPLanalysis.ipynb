| Attribute / Function | Easy Definition                                |
| -------------------- | ---------------------------------------------- |
| `data.shape`         | Returns number of rows and columns             |
| `data.size`          | Returns total number of elements               |
| `data.ndim`          | Returns number of dimensions                   |
| `data.columns`       | Returns column names                           |
| `data.index`         | Returns row index                              |
| `data.dtypes`        | Returns data type of each column               |
| `data.info()`        | Gives summary of dataset                       |
| `data.describe()`    | Gives statistical summary of numerical columns |
| `data.head()`        | Shows first 5 rows                             |
| `data.tail()`        | Shows last 5 rows                              |
| `data.sample()`      | Shows random rows                              |


| Function                | Easy Definition                            |
| ----------------------- | ------------------------------------------ |
| `data["column"]`        | Selects one column                         |
| `data[["col1","col2"]]` | Selects multiple columns                   |
| `data.loc[]`            | Selects rows/columns using labels          |
| `data.iloc[]`           | Selects rows/columns using index positions |
| `data.at[]`             | Accesses a single value using label        |
| `data.iat[]`            | Accesses a single value using position     |
